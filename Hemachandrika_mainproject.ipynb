```python
# STAGE 2 - DATA CLEANING AND PRE-PROCESSING
# CHECK DATA TYPES

print("Data Types:")
print(df.dtypes)
```

```python
# HANDLE DUPLICATES
# Check complete duplicate rows
print("Complete Duplicate Rows:")
print(df.duplicated().sum())
```

```python
# Check duplicate Person_IDs
print("\nDuplicate Person_IDs:")
print(df["Person_ID"].duplicated().sum())
```

```python
# Remove duplicate Person_ID records
df = df.drop_duplicates(subset="Person_ID", keep="first")

print("\nShape After Removing Duplicate Person_IDs:")
print(df.shape)
```

```python
# CLEAN COLUMN NAMES

df.columns = df.columns.str.strip()

print("Column Names:")
print(df.columns.tolist())
```

```python
# Categorical columns
categorical_cols = [
    "Occupation",
    "Area_Selected",
    "Public_Transport",
    "Nearby_Hospitals",
    "Nearby_Schools",
    "Shopping_Access",
    "Noise_Level",
    "Internet_Availability",
    "Main_Reason",
    "Household_Type"
]
```

```python
# Remove spaces and standardize text
for col in categorical_cols:
    df[col] = df[col].str.strip().str.lower()
```

```python
# Convert text to title case
for col in categorical_cols:
    df[col] = df[col].str.title()

print("Categorical values standardized successfully.")
```

```python
# CHECK MISSING VALUES

print("Missing Values:")
print(df.isnull().sum())
```

```python
# CHECK OUTLIERS

numeric_cols = [
    "Monthly_Rent",
    "Distance_To_Work",
    "Safety_Rating",
    "Age",
    "Annual_Income"
]

for col in numeric_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    print(col, ":", len(outliers), "outliers")
```

```python
# Filling negative values as null
df.loc[df["Annual_Income"] < 0, "Annual_Income"] = np.nan
```

```python
# HANDLE OUTLIERS

for col in numeric_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    median_value = df[col].median()

    df.loc[
        (df[col] < lower_limit) |
        (df[col] > upper_limit),
        col
    ] = median_value
```

```python
df["Annual_Income"] = df["Annual_Income"].fillna(
    df["Annual_Income"].median()
)
```

```python
# HANDLE MISSING VALUES

# Fill categorical missing values with mode
for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
# Area-wise median imputation
area_wise_cols = [
    "Monthly_Rent",
    "Distance_To_Work",
    "Safety_Rating"
]

for col in area_wise_cols:
    df[col] = df.groupby("Area_Selected")[col].transform(
        lambda x: x.fillna(x.median())
    )


# Overall median imputation
overall_cols = [
    "Age",
    "Annual_Income"
]

for col in overall_cols:
    df[col] = df[col].fillna(df[col].median())

print("Missing numerical values handled successfully.")
```

```python
# CHECKING NULL

print("Missing Values After Cleaning:")
print(df.isnull().sum())
```

```python
# CHECK SKEWNESS

for col in numeric_cols:
    print(col, ":", df[col].skew())
```

```python
# No transformation required because the numerical variables
# show low to moderate skewness.
print("No transformation required.")
```

```python
# FEATURE TRANSFORMATION

# Calculate monthly income
df["Monthly_Income"] = df["Annual_Income"] / 12

print(df[["Annual_Income", "Monthly_Income"]].head())
```

```python
# Calculate rent-to-income ratio

df["Rent_Income_Ratio"] = (
    df["Monthly_Rent"] / df["Monthly_Income"]
)

print(df[
    ["Monthly_Rent", "Monthly_Income", "Rent_Income_Ratio"]
].head())
```

```python
df["Rent_Income_Ratio"] = pd.to_numeric(
    df["Rent_Income_Ratio"],
    errors="coerce"
)
```

```python
print(df["Rent_Income_Ratio"].dtype)
```

```python
# FINAL DATA QUALITY CHECK

print("Final Dataset Shape:")
print(df.shape)

print("\nFinal Missing Values:")
print(df.isnull().sum())

print("\nFinal Duplicate Rows:")
print(df.duplicated().sum())

print("\nFinal Duplicate Person_IDs:")
print(df["Person_ID"].duplicated().sum())

print("\nFinal Data Types:")
print(df.dtypes)

print("\nFirst 5 Rows of Cleaned Dataset:")
print(df.head())
```
